```python
class Dog:
    species = "Canine" # class attribute

    def __init__(self, name, age):
        self.name = name
        self.age = age

dog1 = Dog("Tommy", 1)
print(dog1.name)
print(dog1.age)
print(dog1.species)
```

```python
dog2 = Dog('David', 2)
print(dog2.name)
print(dog2.age)
print(dog2.species)
```

```python
# __str__method allows us to define a custom string representation of an object
class Dog:
    species = "Canine" # class attribute

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        if self.age <2:
            return f"{self.name} is {self.age} year old."
        return f"{self.name} is {self.age} years old."

dog1 = Dog("Tommy", 1)
dog2 = Dog("David", 2)
print(dog1)
print(dog2)
```

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

car = Car('Honda', 'HRV', 2023)
print(car.make)
print(car.model)
print(car.year)

print('\n')
car2 = Car('Toyota', 'Corolla', 2014)
print(car2.make)
print(car2.model)
print(car2.year)
```

```python
class Car:
    def __init__(this, brand, model):
        this.brand = brand # set instance attribute
        this.model = model

    def display(this):
        return this.brand, this.model

car1 = Car('honda', 'HRV')
print(car1.display())
```

```python
class gfg:
    def __init__(self, topic):
        self._topic = topic

    def topic(self):
        print("Topic:", self._topic) #Access the renamed variable

ins = gfg("Python")
print(ins.topic())
```

```python
# Circle Class for Area Calculation
class Circle:
    def __init__(self, r):
        self.r = r

    def area(self):
        return 3.14 * self.r**2

c1 = Circle(5)

print('Area of the circle:', c1.area())
```
